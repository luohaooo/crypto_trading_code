## Binance USDT-M Websocket 示例

示例：订阅 BTCUSDT 1m kline，并实时打印最新 K 线。

```python

import asyncio
import json
from datetime import datetime

import websockets

# Websocket 端点（USDT-M Futures），直接指定 stream 避免订阅等待
WS_URL = "wss://fstream.binance.com/ws"

symbol = "btcusdt"
timeframe = "1m"
stream = f"{symbol}@kline_{timeframe}"

async def fetch_one_kline():
    ws_url = f"{WS_URL}/{stream}"
    async with websockets.connect(ws_url, ping_interval=20, ping_timeout=10) as ws:
        msg = await asyncio.wait_for(ws.recv(), timeout=5)
        data = json.loads(msg)
        k = data.get("k") or data.get("data", {}).get("k") or data
        if not k:
            print("未获取到 kline 数据")
            return
        open_time = datetime.fromtimestamp(k.get("t", 0) / 1000)
        close_time = datetime.fromtimestamp(k.get("T", 0) / 1000)
        print(
            f"symbol={k.get('s')} open={k.get('o')} high={k.get('h')} "
            f"low={k.get('l')} close={k.get('c')} vol={k.get('v')} "
            f"open_time={open_time} close_time={close_time} is_final={k.get('x')}"
        )


asyncio.run(fetch_one_kline())
```
