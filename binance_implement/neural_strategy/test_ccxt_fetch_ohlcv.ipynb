## Helper Functions

Functions to test fetch_ohlcv and format results.


```python
def test_fetch_ohlcv(exchange, symbol, timeframe, limit, env_name):
    """
    Test fetch_ohlcv function for a given exchange and symbol
    """
    try:
        print(f"\n🔍 Testing {symbol} on {env_name}...")
        
        # Fetch OHLCV data
        ohlcv_data = exchange.fetch_ohlcv(
            symbol=symbol,
            timeframe=timeframe,
            since=since,
            limit=limit
        )
        
        if not ohlcv_data:
            print(f"❌ No data returned for {symbol}")
            return None
        
        # Convert to DataFrame
        df = pd.DataFrame(ohlcv_data, columns=['timestamp', 'open', 'high', 'low', 'close', 'volume'])
        df['datetime'] = pd.to_datetime(df['timestamp'], unit='ms')
        
        print(f"✅ Success: {len(df)} candles fetched")
        print(f"   Time range: {df['datetime'].iloc[0]} to {df['datetime'].iloc[-1]}")
        print(f"   Price range: {df['close'].min():.2f} - {df['close'].max():.2f}")
        print(f"   Volume range: {df['volume'].min():.2f} - {df['volume'].max():.2f}")
        
        return df
        
    except Exception as e:
        print(f"❌ Error fetching {symbol} on {env_name}: {e}")
        return None

def print_dataframe_details(df, symbol, env_name):
    """
    Print detailed DataFrame information
    """
    if df is None:
        return
        
    print(f"\n📊 {symbol} - {env_name} DataFrame Details:")
    print(f"   Shape: {df.shape}")
    print(f"   Columns: {list(df.columns)}")
    print(f"   Data types: {df.dtypes.to_dict()}")
    
    # Show first and last few rows
    print(f"\n📈 First 3 rows:")
    print(df[['datetime', 'open', 'high', 'low', 'close', 'volume']].head(3).to_string(index=False))
    
    print(f"\n📈 Last 3 rows:")
    print(df[['datetime', 'open', 'high', 'low', 'close', 'volume']].tail(3).to_string(index=False))
    
    # Statistical summary
    print(f"\n📊 Statistical Summary:")
    print(df[['open', 'high', 'low', 'close', 'volume']].describe())

print("✅ Helper functions defined")
```

## Testnet Testing

Test fetch_ohlcv function on testnet environment.


```python
print("🔬 Testing TESTNET Environment")
print("=" * 50)

testnet_results = {}

for symbol in test_symbols:
    df = test_fetch_ohlcv(testnet_exchange, symbol, timeframe, limit, "TESTNET")
    if df is not None:
        testnet_results[symbol] = df
        print_dataframe_details(df, symbol, "TESTNET")
    print("-" * 30)

print(f"\n📊 Testnet Summary: {len(testnet_results)}/{len(test_symbols)} symbols successful")
```

## Live Environment Testing

Test fetch_ohlcv function on live environment.


```python
print("🔬 Testing LIVE Environment")
print("=" * 50)

live_results = {}

for symbol in test_symbols:
    df = test_fetch_ohlcv(live_exchange, symbol, timeframe, limit, "LIVE")
    if df is not None:
        live_results[symbol] = df
        print_dataframe_details(df, symbol, "LIVE")
    print("-" * 30)

print(f"\n📊 Live Summary: {len(live_results)}/{len(test_symbols)} symbols successful")
```

## Data Comparison

Compare data between testnet and live environments.


```python
print("🔍 Comparing Testnet vs Live Data")
print("=" * 50)

for symbol in test_symbols:
    if symbol in testnet_results and symbol in live_results:
        testnet_df = testnet_results[symbol]
        live_df = live_results[symbol]
        
        print(f"\n📊 {symbol} Comparison:")
        print(f"   Testnet rows: {len(testnet_df)}")
        print(f"   Live rows: {len(live_df)}")
        
        # Compare latest prices
        testnet_latest = testnet_df.iloc[-1]
        live_latest = live_df.iloc[-1]
        
        print(f"   Testnet latest: {testnet_latest['datetime']} - Close: {testnet_latest['close']:.2f}")
        print(f"   Live latest: {live_latest['datetime']} - Close: {live_latest['close']:.2f}")
        
        # Price difference
        price_diff = abs(testnet_latest['close'] - live_latest['close'])
        price_diff_pct = (price_diff / live_latest['close']) * 100
        
        print(f"   Price difference: {price_diff:.2f} ({price_diff_pct:.4f}%)")
        
        # Time difference
        time_diff = abs(testnet_latest['datetime'] - live_latest['datetime'])
        print(f"   Time difference: {time_diff}")
        
        # Check if data is identical
        if price_diff_pct < 0.01 and time_diff.total_seconds() < 60:  # Less than 0.01% price diff and 1 minute time diff
            print(f"   ✅ Data appears identical")
        else:
            print(f"   ⚠️ Data differs significantly")
            
    elif symbol in testnet_results:
        print(f"\n❌ {symbol}: Only available on testnet")
    elif symbol in live_results:
        print(f"\n❌ {symbol}: Only available on live")
    else:
        print(f"\n❌ {symbol}: Failed on both environments")

print(f"\n🎯 Final Summary:")
print(f"   Testnet successful: {len(testnet_results)}/{len(test_symbols)}")
print(f"   Live successful: {len(live_results)}/{len(test_symbols)}")
print(f"   Both successful: {len(set(testnet_results.keys()) & set(live_results.keys()))}/{len(test_symbols)}")
```

## Performance Testing

Test fetch speed and reliability.


```python
import time

print("⚡ Performance Testing")
print("=" * 50)

def performance_test(exchange, env_name, symbol='BTC/USDT', timeframe='1h', limit=20, iterations=5):
    """
    Test fetch performance
    """
    print(f"\n🏃 Testing {env_name} performance ({iterations} iterations):")
    
    times = []
    successes = 0
    
    for i in range(iterations):
        try:
            start_time = time.time()
            data = exchange.fetch_ohlcv(symbol, timeframe, None, limit)
            end_time = time.time()
            
            fetch_time = end_time - start_time
            times.append(fetch_time)
            successes += 1
            
            print(f"   Iteration {i+1}: {fetch_time:.3f}s ({len(data)} candles)")
            
            # Small delay to avoid rate limiting
            time.sleep(0.1)
            
        except Exception as e:
            print(f"   Iteration {i+1}: Failed - {e}")
    
    if times:
        avg_time = sum(times) / len(times)
        min_time = min(times)
        max_time = max(times)
        
        print(f"\n📊 {env_name} Performance Summary:")
        print(f"   Success rate: {successes}/{iterations} ({successes/iterations*100:.1f}%)")
        print(f"   Average time: {avg_time:.3f}s")
        print(f"   Min time: {min_time:.3f}s")
        print(f"   Max time: {max_time:.3f}s")
        
        return {
            'success_rate': successes/iterations,
            'avg_time': avg_time,
            'min_time': min_time,
            'max_time': max_time
        }
    else:
        print(f"\n❌ {env_name}: All iterations failed")
        return None

# Run performance tests
testnet_perf = performance_test(testnet_exchange, "TESTNET")
live_perf = performance_test(live_exchange, "LIVE")

# Compare performance
if testnet_perf and live_perf:
    print(f"\n🏆 Performance Comparison:")
    print(f"   Testnet avg: {testnet_perf['avg_time']:.3f}s")
    print(f"   Live avg: {live_perf['avg_time']:.3f}s")
    
    if testnet_perf['avg_time'] < live_perf['avg_time']:
        diff = live_perf['avg_time'] - testnet_perf['avg_time']
        print(f"   🥇 Testnet is faster by {diff:.3f}s")
    else:
        diff = testnet_perf['avg_time'] - live_perf['avg_time']
        print(f"   🥇 Live is faster by {diff:.3f}s")

print("\n✅ Performance testing completed")
```
